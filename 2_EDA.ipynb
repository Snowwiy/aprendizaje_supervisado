# Importar y cargar datos

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("./data/titanic_clean.csv")
df.head()
```

# Revisión general

```python
print("Dimensiones del dataset:", df.shape)
print("\nInformación general:")
df.info()

print("\nEstadísticas descriptivas:")
display(df.describe())
```

# Gráficas principales

```python
sns.countplot(x="Survived", data=df)
plt.title("Distribución de Supervivencia")
plt.show()

sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Supervivencia por Sexo")
plt.show()

sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Supervivencia por Clase")
plt.show()

sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Distribución de Edad")
plt.show()
```

# Correlación y conclusión visual

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Mapa de Correlación")
plt.show()

sns.boxplot(x="Survived", y="Fare", data=df)
plt.title("Tarifa vs Supervivencia")
plt.show()
```

## Conclusión

El análisis exploratorio muestra que variables como Sex, Pclass, Age y Fare pueden influir en la supervivencia. Estos patrones serán útiles para preparar las variables y entrenar modelos de machine learning.
